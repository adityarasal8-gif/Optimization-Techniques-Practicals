```python
import sympy as sp
```

```python
# Define variables
x, lam=sp.symbols('x lam',nonnegative=True)

# Objective function
f=x**2

# Constraint: x >= 2
g=2-x

# Lagrangian
L=f+lam*g

# Stationarity condition
eql= sp.diff(L,x)

# Complementary slackness
eq2= lam*g

# Solve equations
solutions= sp.solve([eql,eq2],(x,lam),dict= True)
print("Possible Solutions : ")
print(solutions)

for sol in solutions:
  x_val=sol[x]
  lam_val=sol[lam]

  if x_val>=2 and lam_val>=0:
    print("\nOptimal Solution")
    print("x=",x_val)
    print("lambda=",lam_val)
    print("Minimum Value=",f.subs(x,x_val))
```
